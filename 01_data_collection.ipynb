```python
# Cell 2 - Load and inspect crop production data
crop_df = pd.read_csv(f'{RAW}\\crop_production.csv')

print("Shape:", crop_df.shape)
print("\nColumns:", crop_df.columns.tolist())
print("\nFirst 5 rows:")
crop_df.head()
```

```python
# Cell 3 - Check basic info
print("Data Types:")
print(crop_df.dtypes)
print("\nMissing Values:")
print(crop_df.isnull().sum())
print("\nUnique States:")
print(crop_df['State'].unique())
```

```python
# Cell 4 - Filter Maharashtra only
maha_df = crop_df[crop_df['State'].str.upper() == 'MAHARASHTRA'].copy()

print("Maharashtra records:", len(maha_df))
print("Districts:", maha_df['District'].nunique())
print("Crops:", maha_df['Crop'].nunique())
print("Year range:", maha_df['Year'].min(), "to", maha_df['Year'].max())
print("\nDistricts list:")
print(sorted(maha_df['District'].unique()))
```

```python
# Cell 5 - Rename columns and calculate yield
maha_df = maha_df.rename(columns={
    'State': 'state',
    'District': 'district',
    'Year': 'year',
    'Season': 'season',
    'Crop': 'crop',
    'Area': 'area_hectares',
    'Production': 'production_tonnes',
    'Yield': 'yield_kg_per_hectare'
})

# Remove bad rows
maha_df = maha_df[maha_df['area_hectares'] > 0]
maha_df = maha_df[maha_df['production_tonnes'] >= 0]
maha_df = maha_df[maha_df['yield_kg_per_hectare'] < 50000]
maha_df = maha_df.dropna(subset=['crop', 'season', 'area_hectares'])

print("After cleaning shape:", maha_df.shape)
print("\nSample:")
maha_df.head()
```

```python
# Cell 6 - Filter top Maharashtra crops
top_crops = maha_df['crop'].value_counts().head(20)
print("Top 20 crops by records:")
print(top_crops)
```

```python
# Cell 10 - Filter all Maharashtra subdivisions
maha_subdivisions = [
    'Konkan & Goa',
    'Madhya Maharashtra', 
    'Matathwada',
    'Vidarbha'
]

rain_maha = rain_df[rain_df['SUBDIVISION'].isin(maha_subdivisions)].copy()

print("Shape:", rain_maha.shape)
print("\nSubdivisions:")
print(rain_maha['SUBDIVISION'].unique())
print("\nYear range:", rain_maha['YEAR'].min(), "to", rain_maha['YEAR'].max())
print("\nSample:")
rain_maha[['SUBDIVISION', 'YEAR', 'JUN', 'JUL', 'AUG', 'SEP', 'ANNUAL']].head()
```

```python
# Cell 11 - Clean rainfall and calculate monsoon + deviation
rain_maha = rain_maha.rename(columns={
    'SUBDIVISION': 'subdivision',
    'YEAR': 'year',
    'JUN': 'rain_june',
    'JUL': 'rain_july',
    'AUG': 'rain_august',
    'SEP': 'rain_september',
    'ANNUAL': 'annual_rainfall'
})

# Calculate monsoon rainfall
rain_maha['monsoon_rainfall'] = (
    rain_maha['rain_june'] +
    rain_maha['rain_july'] +
    rain_maha['rain_august'] +
    rain_maha['rain_september']
)

# Calculate 30 year normal per subdivision
normal_rain = rain_maha.groupby('subdivision')['annual_rainfall'].mean().reset_index()
normal_rain.columns = ['subdivision', 'normal_annual_rainfall']

# Merge back
rain_maha = rain_maha.merge(normal_rain, on='subdivision', how='left')

# Calculate deviation
rain_maha['rainfall_deviation_pct'] = (
    (rain_maha['annual_rainfall'] - rain_maha['normal_annual_rainfall'])
    / rain_maha['normal_annual_rainfall'] * 100
).round(2)

print("Shape:", rain_maha.shape)
print("\nSample:")
rain_maha[['subdivision', 'year', 'annual_rainfall',
           'monsoon_rainfall', 'rainfall_deviation_pct']].head()
```

```python
# Cell 13 - Merge crop data with rainfall data
# First filter rainfall to match our crop year range (1997-2019)
rain_maha['year'] = rain_maha['year'].astype(int)

# Extract year from crop data (1997-98 -> 1997)
maha_crops_df['year_num'] = maha_crops_df['year'].str[:4].astype(int)

# Merge
master_df = maha_crops_df.merge(
    rain_maha[['subdivision', 'year', 'annual_rainfall',
               'monsoon_rainfall', 'rainfall_deviation_pct',
               'rain_june', 'rain_july', 'rain_august', 'rain_september',
               'normal_annual_rainfall']],
    left_on=['subdivision', 'year_num'],
    right_on=['subdivision', 'year'],
    how='left'
)

print("Master dataframe shape:", master_df.shape)
print("\nMissing values:")
print(master_df.isnull().sum())
print("\nSample:")
master_df[['district', 'crop', 'year_x', 'yield_kg_per_hectare',
           'annual_rainfall', 'rainfall_deviation_pct']].head()
```

```python
# Cell 18 - Smart outlier removal per crop
maha_crops_df = maha_crops_df.reset_index(drop=True)

def remove_outliers(group):
    Q1 = group['yield_kg_per_hectare'].quantile(0.05)
    Q3 = group['yield_kg_per_hectare'].quantile(0.95)
    return group[
        (group['yield_kg_per_hectare'] >= Q1) &
        (group['yield_kg_per_hectare'] <= Q3)
    ]

# Save crop column separately before groupby
crop_col = maha_crops_df['crop'].copy()

# Apply outlier removal
clean_parts = []
for crop_name in maha_crops_df['crop'].unique():
    subset = maha_crops_df[maha_crops_df['crop'] == crop_name].copy()
    cleaned = remove_outliers(subset)
    clean_parts.append(cleaned)

maha_crops_clean = pd.concat(clean_parts).reset_index(drop=True)

print("Before:", maha_crops_df.shape)
print("After:", maha_crops_clean.shape)
print("Columns:", maha_crops_clean.columns.tolist())
print("\nSample yields:")
for crop in maha_crops_clean['crop'].unique():
    subset = maha_crops_clean[maha_crops_clean['crop'] == crop]['yield_kg_per_hectare']
    print(f"{crop}: min={subset.min():.0f} mean={subset.mean():.0f} max={subset.max():.0f}")
```
